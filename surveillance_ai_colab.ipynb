## 🚀 Step 6: Start the FastAPI Server

```python
# Get the public URL for the server
from google.colab.output import eval_js
import uvicorn
import threading
import time

# Get the public URL
public_url = eval_js("google.colab.kernel.proxyPort(8000)")
print(f"🌐 Server will be available at: {public_url}")
print(f"📊 API Documentation: {public_url}/docs")
print(f"🔍 Health Check: {public_url}/health")

# Start the server in a separate thread
def start_server():
    try:
        uvicorn.run("main:app", host="0.0.0.0", port=8000, log_level="info")
    except Exception as e:
        print(f"❌ Server failed to start: {e}")

print("\n🚀 Starting FastAPI server...")
server_thread = threading.Thread(target=start_server)
server_thread.daemon = True
server_thread.start()

# Wait a moment for server to start
time.sleep(5)
print("✅ Server started successfully!")
print(f"\n🎯 Your Surveillance AI Platform is now running!")
print(f"📱 Use the URL above to access your API")
```

## 🧪 Step 7: Test the API

```python
# Test the API endpoints using the internal localhost address for reliability
import requests
import json
import time

# For internal testing within the Colab notebook, we use localhost.
# The public_url is for you to use from your browser or an external API client.
internal_url = "http://localhost:8000"

print(f"🧪 Starting API tests against internal URL: {internal_url}")
print(f"🌐 Your PUBLIC URL for external access is still: {public_url}")

# --- Health Check ---
health_check_url = f"{internal_url}/health"
health_check_passed = False

# Give the server a moment to start up fully
time.sleep(5)

try:
    response = requests.get(health_check_url, timeout=10)
    if response.status_code == 200:
        print("\n✅ Health check passed!", response.json())
        health_check_passed = True
    else:
        print(f"\n❌ Health check failed with status {response.status_code}: {response.text}")
except requests.exceptions.RequestException as e:
    print(f"\n❌ Health check request failed: {e}")

# --- API Test Functions ---
def test_video_analysis(video_path, prompt, model="chatgpt"):
    """Test the video analysis endpoint using the internal URL"""
    try:
        # We use the internal_url for the test script
        url = f"{internal_url}/api/v1/analyze"
        with open(video_path, "rb") as video_file:
            files = {"video": video_file}
            data = {"prompt": prompt, "model": model}

            print(f"\n🚀 Sending request to {url}...")
            # Increased timeout for potentially long analysis
            response = requests.post(url, files=files, data=data, timeout=180)
            print("✅ Response received!")
            return response.json()
    except Exception as e:
        return {"error": str(e)}

print("\n✅ API test functions ready")
print("📝 To test video analysis, use:")
print("   test_video_analysis('path/to/video.mp4', 'your prompt')")

if health_check_passed:
    print("\n🎉 Setup complete! Your Surveillance AI Platform is ready to use.")
else:
    print("\n⚠️ API may not be fully functional. Please check server logs in Cell 6.")
```
